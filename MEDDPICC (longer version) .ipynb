```python
from typing import Dict, List, Tuple
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from datetime import datetime

class EnhancedMEDDPICCScoring:
    def __init__(self):
        self.base_weights = {
            'Metrics': {
                'weight': 0.20,
                'min_threshold': 0.6,
                'critical': True
            },
            'Economic_Buyer': {
                'weight': 0.15,
                'min_threshold': 0.7,
                'critical': True
            },
            'Decision_Criteria': {
                'weight': 0.10,
                'min_threshold': 0.5,
                'critical': False
            },
            'Decision_Process': {
                'weight': 0.15,
                'min_threshold': 0.6,
                'critical': True
            },
            'Paper_Process': {
                'weight': 0.05,
                'min_threshold': 0.4,
                'critical': False
            },
            'Identified_Pain': {
                'weight': 0.15,
                'min_threshold': 0.6,
                'critical': True
            },
            'Champion': {
                'weight': 0.15,
                'min_threshold': 0.6,
                'critical': True
            },
            'Competition': {
                'weight': 0.05,
                'min_threshold': 0.4,
                'critical': False
            }
        }

        self.ml_model = RandomForestClassifier(n_estimators=100, random_state=42)
        self.historical_data = []

    def detailed_scoring_criteria(self):
        """Enhanced scoring criteria with multiple levels and validations"""
        return {
            'Metrics': {
                'quantification': {
                    'weight': 0.4,
                    'components': {
                        'roi_defined': {
                            'weight': 0.3,
                            'validation': ['calculation_method', 'assumptions_documented'],
                            'evidence_required': True
                        },
                        'impact_calculated': {
                            'weight': 0.4,
                            'validation': ['before_after_metrics', 'timeline_defined'],
                            'evidence_required': True
                        },
                        'baseline_established': {
                            'weight': 0.3,
                            'validation': ['current_metrics', 'data_verified'],
                            'evidence_required': True
                        }
                    }
                },
                'validation': {
                    'weight': 0.3,
                    'components': {
                        'buyer_confirmed': {
                            'weight': 0.5,
                            'validation': ['written_confirmation', 'meeting_documented'],
                            'evidence_required': True
                        },
                        'data_verified': {
                            'weight': 0.5,
                            'validation': ['source_identified', 'methodology_confirmed'],
                            'evidence_required': True
                        }
                    }
                },
                'documentation': {
                    'weight': 0.3,
                    'components': {
                        'formal_documentation': {
                            'weight': 0.6,
                            'validation': ['document_exists', 'stakeholder_reviewed'],
                            'evidence_required': True
                        },
                        'sharing_approved': {
                            'weight': 0.4,
                            'validation': ['approval_received', 'distribution_list'],
                            'evidence_required': False
                        }
                    }
                }
            }
        }

    def calculate_component_score(self, data: Dict, criteria: Dict) -> Tuple[float, List[str]]:
        """Calculate score for individual components with validation"""
        score = 0
        missing_elements = []

        for component, details in criteria['components'].items():
            component_score = 0
            for validation_item in details['validation']:
                if validation_item in data and data[validation_item]:
                    component_score += 1
                elif details['evidence_required']:
                    missing_elements.append(f"{component}: {validation_item}")

            normalized_score = (component_score / len(details['validation'])) * details['weight']
            score += normalized_score

        return score, missing_elements

    def apply_dynamic_weights(self, base_score: float, context: Dict) -> float:
        """Apply dynamic weights based on context"""
        modifiers = {
            'deal_size': {
                'enterprise': 1.2,
                'mid_market': 1.0,
                'small': 0.8
            },
            'sales_cycle': {
                'early': 0.9,
                'middle': 1.0,
                'late': 1.1
            },
            'competition': {
                'high': 1.2,
                'medium': 1.0,
                'low': 0.8
            }
        }

        adjusted_score = base_score
        for factor, value in context.items():
            if factor in modifiers and value in modifiers[factor]:
                adjusted_score *= modifiers[factor][value]

        return min(adjusted_score, 100)  # Cap at 100

    def calculate_risk_score(self, data: Dict) -> Dict:
        """Calculate risk score and identify risk factors"""
        risk_factors = {
            'critical_elements_missing': [],
            'validation_gaps': [],
            'timeline_risks': [],
            'engagement_risks': []
        }

        risk_score = 100

        for category, details in self.base_weights.items():
            if details['critical']:
                category_score = self.calculate_category_score(category, data)
                if category_score < details['min_threshold']:
                    risk_factors['critical_elements_missing'].append(category)
                    risk_score -= 20

        for category, score in data.get('validation_scores', {}).items():
            if score < 0.5:
                risk_factors['validation_gaps'].append(category)
                risk_score -= 10

        return {
            'risk_score': max(risk_score, 0),
            'risk_factors': risk_factors,
            'recommendations': self.generate_risk_recommendations(risk_factors)
        }

    def predict_outcome_probability(self, data: Dict) -> float:
        """Use ML model to predict deal outcome probability"""
        if len(self.historical_data) > 100:  # Minimum data for ML
            features = self.extract_features(data)
            probability = self.ml_model.predict_proba([features])[0][1]
            return probability
        return None

    def calculate_momentum_score(self, historical_scores: List[Dict]) -> Dict:
        """Calculate deal momentum based on score progression"""
        if len(historical_scores) < 2:
            return {'momentum': 0, 'trend': 'neutral'}

        scores = [score['overall_score'] for score in historical_scores]
        momentum = np.mean(np.diff(scores))

        return {
            'momentum': momentum,
            'trend': 'positive' if momentum > 0 else 'negative',
            'velocity': abs(momentum),
            'consistency': np.std(scores)
        }

    def generate_coaching_insights(self, data: Dict) -> Dict:
        """Generate coaching insights based on scoring analysis"""
        insights = {
            'strengths': [],
            'gaps': [],
            'next_actions': [],
            'risk_mitigations': []
        }

        for category, score in data['category_scores'].items():
            if score['raw_score'] >= 0.8:
                insights['strengths'].append(category)
            elif score['raw_score'] <= 0.6:
                insights['gaps'].append(category)
                insights['next_actions'].append(
                    self.get_improvement_actions(category, score['raw_score'])
                )

        return insights

    def calculate_final_score(self, data: Dict, context: Dict) -> Dict:
        """Calculate final MEDDPICC score with all enhancements"""
        category_scores = {}
        missing_elements = []
        overall_score = 0

        for category, weight_info in self.base_weights.items():
            category_score, category_missing = self.calculate_category_score(
                category, data)
            weighted_score = category_score * weight_info['weight']

            category_scores[category] = {
                'raw_score': category_score,
                'weighted_score': weighted_score,
                'missing_elements': category_missing
            }

            overall_score += weighted_score
            missing_elements.extend(category_missing)

        adjusted_score = self.apply_dynamic_weights(overall_score, context)

        risk_analysis = self.calculate_risk_score(data)
        momentum = self.calculate_momentum_score(data.get('historical_scores', []))
        prediction = self.predict_outcome_probability(data)
        coaching = self.generate_coaching_insights(
            {'category_scores': category_scores})

        return {
            'overall_score': adjusted_score,
            'category_scores': category_scores,
            'missing_elements': missing_elements,
            'risk_analysis': risk_analysis,
            'momentum': momentum,
            'win_probability': prediction,
            'coaching_insights': coaching,
            'last_updated': datetime.now(),
            'confidence_level': self.calculate_confidence_level(data)
        }

    def calculate_confidence_level(self, data: Dict) -> float:
        """Calculate confidence level in the scoring"""
        factors = {
            'data_completeness': len(data) / len(self.detailed_scoring_criteria()),
            'evidence_quality': sum(1 for x in data.values() if x.get('evidence_provided', False)) / len(data),
            'validation_level': sum(1 for x in data.values() if x.get('validated', False)) / len(data),
            'time_relevance': 1.0  # Assume current data
        }

        return np.mean(list(factors.values())) * 100
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

def generate_test_dataset(n_opportunities=1000):
    """
    Generate synthetic dataset combining AQL signals and MEDDPICC elements
    """

    def generate_company_size():
        return np.random.choice(
            ['Enterprise', 'Mid-Market', 'SMB'],
            p=[0.3, 0.4, 0.3]
        )

    def generate_deal_size(company_size):
        if company_size == 'Enterprise':
            return random.uniform(250000, 1000000)
        elif company_size == 'Mid-Market':
            return random.uniform(100000, 250000)
        else:
            return random.uniform(25000, 100000)

    def generate_intent_signals():
        return random.uniform(0, 100)

    def generate_meddpicc_scores():
        return {
            'metrics_score': random.uniform(0, 100),
            'economic_buyer_score': random.uniform(0, 100),
            'decision_criteria_score': random.uniform(0, 100),
            'decision_process_score': random.uniform(0, 100),
            'paper_process_score': random.uniform(0, 100),
            'identified_pain_score': random.uniform(0, 100),
            'champion_score': random.uniform(0, 100),
            'competition_score': random.uniform(0, 100)
        }

    opportunities = []

    for i in range(n_opportunities):
        company_size = generate_company_size()
        deal_size = generate_deal_size(company_size)

        aql_signals = {
            'intent_score': generate_intent_signals(),
            'tech_stack_changes': random.randint(0, 5),
            'competitor_research_count': random.randint(0, 20),
            'buying_committee_size': random.randint(2, 10),
            'relevant_job_postings': random.randint(0, 15),
            'website_engagement_score': random.uniform(0, 100),
            'content_engagement_score': random.uniform(0, 100)
        }

        meddpicc_scores = generate_meddpicc_scores()

        engagement_metrics = {
            'email_opens': random.randint(0, 50),
            'email_replies': random.randint(0, 20),
            'meetings_held': random.randint(0, 10),
            'document_views': random.randint(0, 30),
            'time_spent_on_site': random.randint(0, 3600)
        }

        deal_characteristics = {
            'sales_cycle_days': random.randint(30, 180),
            'number_of_stakeholders': random.randint(2, 15),
            'technical_complexity': random.uniform(0, 100),
            'competitive_deal': random.choice([True, False]),
            'budget_identified': random.choice([True, False])
        }

        opportunity = {
            'opportunity_id': f'OPP-{i+1:04d}',
            'company_size': company_size,
            'deal_size': deal_size,
            'created_date': (datetime.now() - timedelta(days=random.randint(0, 180))).strftime('%Y-%m-%d'),
            'stage': np.random.choice(
                ['Discovery', 'Evaluation', 'Proposal', 'Negotiation'],
                p=[0.3, 0.3, 0.2, 0.2]
            ),

            **aql_signals,

            **meddpicc_scores,

            **engagement_metrics,

            **deal_characteristics,

            'won': random.choice([True, False]),
            'closed_amount': deal_size if random.choice([True, False]) else 0,
            'days_to_close': random.randint(30, 180)
        }

        if (opportunity['intent_score'] > 70 and
            opportunity['metrics_score'] > 70 and
            opportunity['economic_buyer_score'] > 70):
            opportunity['won'] = random.choice([True, True, False])  # 66% chance of winning

        if opportunity['competitive_deal'] and opportunity['competition_score'] < 50:
            opportunity['won'] = random.choice([True, False, False])  # 33% chance of winning

        opportunities.append(opportunity)

    df = pd.DataFrame(opportunities)

    df['total_engagement_score'] = (
        df['email_opens'] * 0.2 +
        df['email_replies'] * 0.3 +
        df['meetings_held'] * 0.3 +
        df['document_views'] * 0.2
    )

    df['meddpicc_total_score'] = (
        df['metrics_score'] * 0.2 +
        df['economic_buyer_score'] * 0.15 +
        df['decision_criteria_score'] * 0.1 +
        df['decision_process_score'] * 0.15 +
        df['paper_process_score'] * 0.05 +
        df['identified_pain_score'] * 0.15 +
        df['champion_score'] * 0.15 +
        df['competition_score'] * 0.05
    )

    df['aql_quality_score'] = (
        df['intent_score'] * 0.4 +
        df['website_engagement_score'] * 0.3 +
        df['content_engagement_score'] * 0.3
    )

    df = df.apply(lambda x: x + np.random.normal(0, 0.1) if np.issubdtype(x.dtype, np.number) else x)

    return df

test_data = generate_test_dataset(1000)

print("\nDataset Summary:")
print(f"Total Opportunities: {len(test_data)}")
print(f"Win Rate: {test_data['won'].mean():.1%}")
print(f"Average Deal Size: ${test_data['deal_size'].mean():,.2f}")
print(f"Average Sales Cycle: {test_data['sales_cycle_days'].mean():.1f} days")

meddpicc_columns = [col for col in test_data.columns if '_score' in col]
correlations = test_data[meddpicc_columns + ['won']].corr()['won'].sort_values(ascending=False)

print("\nMEDDPICC Score Correlations with Win Rate:")
print(correlations)

test_data.to_csv('aql_meddpicc_test_data.csv', index=False)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

class ProspectoroAnalytics:
    def __init__(self, data):
        self.data = data
        # Set basic figure style
        plt.rcParams['figure.figsize'] = [15, 10]
        plt.rcParams['figure.dpi'] = 100

    def create_summary_report(self):
        """Generate basic analysis report"""
        # Create figure with subplots
        fig, axs = plt.subplots(3, 2, figsize=(20, 25))
        fig.suptitle('Prospectoro Analytics Dashboard', fontsize=16)

        # 1. Win Rate Analysis
        self.plot_win_rate_analysis(axs[0, 0])

        # 2. Deal Size Analysis
        self.plot_deal_size_analysis(axs[0, 1])

        # 3. MEDDPICC Analysis
        self.plot_meddpicc_analysis(axs[1, 0])

        # 4. AQL Analysis
        self.plot_aql_analysis(axs[1, 1])

        # 5. Sales Cycle Analysis
        self.plot_sales_cycle_analysis(axs[2, 0])

        # 6. Print Summary Stats
        self.plot_summary_stats(axs[2, 1])

        plt.tight_layout()
        return fig

    def plot_win_rate_analysis(self, ax):
        """Plot win rate by company size"""
        win_rates = self.data.groupby('company_size')['won'].mean()
        win_rates.plot(kind='bar', ax=ax)
        ax.set_title('Win Rate by Company Size')
        ax.set_ylabel('Win Rate')
        ax.set_ylim(0, 1)

        # Add percentage labels
        for i, v in enumerate(win_rates):
            ax.text(i, v + 0.01, f'{v:.1%}', ha='center')

    def plot_deal_size_analysis(self, ax):
        """Plot average deal size by company size"""
        avg_deals = self.data.groupby('company_size')['deal_size'].mean()
        avg_deals.plot(kind='bar', ax=ax)
        ax.set_title('Average Deal Size by Company Size')
        ax.set_ylabel('Deal Size ($)')

        # Add value labels
        for i, v in enumerate(avg_deals):
            ax.text(i, v, f'${v:,.0f}', ha='center', va='bottom')

    def plot_meddpicc_analysis(self, ax):
        """Plot MEDDPICC score comparison"""
        meddpicc_cols = [col for col in self.data.columns if '_score' in col]

        # Calculate average scores
        won_scores = self.data[self.data['won']][meddpicc_cols].mean()
        lost_scores = self.data[~self.data['won']][meddpicc_cols].mean()

        x = np.arange(len(meddpicc_cols))
        width = 0.35

        ax.bar(x - width/2, won_scores, width, label='Won')
        ax.bar(x + width/2, lost_scores, width, label='Lost')

        ax.set_title('MEDDPICC Scores: Won vs Lost')
        ax.set_xticks(x)
        ax.set_xticklabels([col.replace('_score', '') for col in meddpicc_cols],
                          rotation=45)
        ax.legend()

    def plot_aql_analysis(self, ax):
        """Plot AQL intent score analysis"""
        won_intent = self.data[self.data['won']]['intent_score']
        lost_intent = self.data[~self.data['won']]['intent_score']

        ax.hist([won_intent, lost_intent], label=['Won', 'Lost'], bins=20)
        ax.set_title('Intent Score Distribution')
        ax.set_xlabel('Intent Score')
        ax.legend()

    def plot_sales_cycle_analysis(self, ax):
        """Plot sales cycle analysis"""
        avg_cycle = self.data.groupby('stage')['sales_cycle_days'].mean()
        avg_cycle.plot(kind='bar', ax=ax)
        ax.set_title('Average Sales Cycle by Stage')
        ax.set_ylabel('Days')

        # Add value labels
        for i, v in enumerate(avg_cycle):
            ax.text(i, v, f'{v:.0f}', ha='center', va='bottom')

    def plot_summary_stats(self, ax):
        """Plot summary statistics"""
        ax.axis('off')
        summary_text = f"""
        SUMMARY STATISTICS

        Total Opportunities: {len(self.data):,}
        Total Pipeline: ${self.data['deal_size'].sum():,.0f}
        Average Deal: ${self.data['deal_size'].mean():,.0f}
        Win Rate: {self.data['won'].mean():.1%}
        Avg Sales Cycle: {self.data['sales_cycle_days'].mean():.0f} days
        """
        ax.text(0.1, 0.5, summary_text, fontsize=12)

    def print_detailed_analysis(self):
        """Print detailed statistical analysis"""
        print("\nPROSPECTORO ANALYTICS REPORT")
        print("=" * 50)

        print("\nConversion Rates by Company Size:")
        print("-" * 30)
        for size in self.data['company_size'].unique():
            size_data = self.data[self.data['company_size'] == size]
            print(f"{size}: {size_data['won'].mean():.1%}")

        print("\nMEDDPICC Score Analysis:")
        print("-" * 30)
        meddpicc_cols = [col for col in self.data.columns if '_score' in col]
        for col in meddpicc_cols:
            won_avg = self.data[self.data['won']][col].mean()
            lost_avg = self.data[~self.data['won']][col].mean()
            print(f"\n{col}:")
            print(f"  Won deals avg: {won_avg:.1f}")
            print(f"  Lost deals avg: {lost_avg:.1f}")

        print("\nAQL Analysis:")
        print("-" * 30)
        print(f"Average Intent Score: {self.data['intent_score'].mean():.1f}")
        print(f"Intent Score (Won): {self.data[self.data['won']]['intent_score'].mean():.1f}")
        print(f"Intent Score (Lost): {self.data[~self.data['won']]['intent_score'].mean():.1f}")

# Usage
if __name__ == "__main__":
    # Assuming test_data exists from previous code
    analytics = ProspectoroAnalytics(test_data)

    # Generate visual report
    fig = analytics.create_summary_report()

    # Save the report
    plt.savefig('prospectoro_report.png', bbox_inches='tight')
    plt.close()

    # Print detailed analysis
    analytics.print_detailed_analysis()
```

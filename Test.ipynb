```python
class LeadClassifier:
    def __init__(self):
        self.classification_thresholds = {
            'hot_lead': 0.8,
            'warm_lead': 0.5,
            'cold_lead': 0.2
        }

    async def classify_lead(self, lead_data: dict):
        scores = {
            'engagement_score': self.calculate_engagement_score(
                lead_data['engagement_signals']
            ),
            'intent_score': self.calculate_intent_score(
                lead_data['intent_signals']
            ),
            'firmographic_score': self.calculate_firmographic_score(
                lead_data['company_data']
            )
        }

        final_score = self.calculate_final_score(scores)

        return self.get_classification(final_score)

    def get_classification(self, score: float):
        if score >= self.classification_thresholds['hot_lead']:
            return 'hot_lead'
        elif score >= self.classification_thresholds['warm_lead']:
            return 'warm_lead'
        elif score >= self.classification_thresholds['cold_lead']:
            return 'cold_lead'
        else:
            return 'not_qualified'
```

```python
class ClassificationEvaluator:
    def evaluate_classifications(self, predictions: list, actual_outcomes: list):
        metrics = {
            'accuracy': self.calculate_accuracy(predictions, actual_outcomes),
            'precision': self.calculate_precision(predictions, actual_outcomes),
            'recall': self.calculate_recall(predictions, actual_outcomes),
            'f1_score': self.calculate_f1_score(predictions, actual_outcomes),
            'confusion_matrix': self.generate_confusion_matrix(
                predictions,
                actual_outcomes
            )
        }

        return ClassificationReport(
            metrics=metrics,
            detailed_analysis=self.analyze_results(metrics),
            recommendations=self.generate_recommendations(metrics)
        )

    def analyze_results(self, metrics: dict):
        return {
            'classification_distribution': self.analyze_distribution(metrics),
            'error_analysis': self.analyze_errors(metrics),
            'performance_by_category': self.analyze_category_performance(metrics)
        }

    def calculate_accuracy(self, predictions: list, actual_outcomes: list):
        correct = sum(1 for p, a in zip(predictions, actual_outcomes) if p == a)
        total = len(predictions)
        return correct / total if total > 0 else 0

    def calculate_precision(self, predictions: list, actual_outcomes: list):
        true_positives = sum(1 for p, a in zip(predictions, actual_outcomes) if p == a and p == 1)
        predicted_positives = sum(1 for p in predictions if p == 1)
        return true_positives / predicted_positives if predicted_positives > 0 else 0

    def calculate_recall(self, predictions: list, actual_outcomes: list):
        true_positives = sum(1 for p, a in zip(predictions, actual_outcomes) if p == a and p == 1)
        actual_positives = sum(1 for a in actual_outcomes if a == 1)
        return true_positives / actual_positives if actual_positives > 0 else 0

    def calculate_f1_score(self, predictions: list, actual_outcomes: list):
        precision = self.calculate_precision(predictions, actual_outcomes)
        recall = self.calculate_recall(predictions, actual_outcomes)
        return 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    def generate_confusion_matrix(self, predictions: list, actual_outcomes: list):
        tp = sum(1 for p, a in zip(predictions, actual_outcomes) if p == a and p == 1)
        tn = sum(1 for p, a in zip(predictions, actual_outcomes) if p == a and p == 0)
        fp = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 1 and a == 0)
        fn = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 0 and a == 1)

        return {
            'true_positives': tp,
            'true_negatives': tn,
            'false_positives': fp,
            'false_negatives': fn
        }

    def analyze_distribution(self, metrics: dict):
        confusion_matrix = metrics['confusion_matrix']
        total = sum(confusion_matrix.values())

        return {
            'positive_rate': (confusion_matrix['true_positives'] +
                            confusion_matrix['false_positives']) / total,
            'negative_rate': (confusion_matrix['true_negatives'] +
                            confusion_matrix['false_negatives']) / total,
            'class_distribution': {
                'positive': (confusion_matrix['true_positives'] +
                           confusion_matrix['false_negatives']) / total,
                'negative': (confusion_matrix['true_negatives'] +
                           confusion_matrix['false_positives']) / total
            }
        }

    def analyze_errors(self, metrics: dict):
        confusion_matrix = metrics['confusion_matrix']
        total_errors = confusion_matrix['false_positives'] + confusion_matrix['false_negatives']
        total_samples = sum(confusion_matrix.values())

        return {
            'error_rate': total_errors / total_samples if total_samples > 0 else 0,
            'false_positive_rate': (confusion_matrix['false_positives'] /
                                  (confusion_matrix['false_positives'] +
                                   confusion_matrix['true_negatives'])
                                  if (confusion_matrix['false_positives'] +
                                      confusion_matrix['true_negatives']) > 0 else 0),
            'false_negative_rate': (confusion_matrix['false_negatives'] /
                                  (confusion_matrix['false_negatives'] +
                                   confusion_matrix['true_positives'])
                                  if (confusion_matrix['false_negatives'] +
                                      confusion_matrix['true_positives']) > 0 else 0)
        }

    def analyze_category_performance(self, metrics: dict):
        return {
            'overall_accuracy': metrics['accuracy'],
            'class_metrics': {
                'positive': {
                    'precision': metrics['precision'],
                    'recall': metrics['recall'],
                    'f1_score': metrics['f1_score']
                }
            }
        }

    def generate_recommendations(self, metrics: dict):
        recommendations = []
        error_analysis = self.analyze_errors(metrics)

        if error_analysis['false_positive_rate'] > 0.2:
            recommendations.append("Consider adjusting threshold to reduce false positives")

        if error_analysis['false_negative_rate'] > 0.2:
            recommendations.append("Consider adjusting threshold to reduce false negatives")

        if metrics['accuracy'] < 0.8:
            recommendations.append("Model performance needs improvement - consider retraining")

        return recommendations
```

```python
from dataclasses import dataclass
from typing import Dict, List

@dataclass
class ClassificationReport:
    metrics: Dict
    detailed_analysis: Dict
    recommendations: List[str]

class ClassificationEvaluator:
    def evaluate_classifications(self, predictions: list, actual_outcomes: list):
        metrics = {
            'accuracy': self.calculate_accuracy(predictions, actual_outcomes),
            'precision': self.calculate_precision(predictions, actual_outcomes),
            'recall': self.calculate_recall(predictions, actual_outcomes),
            'f1_score': self.calculate_f1_score(predictions, actual_outcomes),
            'confusion_matrix': self.generate_confusion_matrix(
                predictions,
                actual_outcomes
            )
        }

        return ClassificationReport(
            metrics=metrics,
            detailed_analysis=self.analyze_results(metrics),
            recommendations=self.generate_recommendations(metrics)
        )

    def calculate_accuracy(self, predictions: list, actual_outcomes: list) -> float:
        if len(predictions) != len(actual_outcomes):
            raise ValueError("Predictions and actual outcomes must be the same length")
        correct = sum(1 for p, a in zip(predictions, actual_outcomes) if p == a)
        return correct / len(predictions) if len(predictions) > 0 else 0.0

    def calculate_precision(self, predictions: list, actual_outcomes: list) -> float:
        true_positives = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 1 and a == 1)
        predicted_positives = sum(1 for p in predictions if p == 1)
        return true_positives / predicted_positives if predicted_positives > 0 else 0.0

    def calculate_recall(self, predictions: list, actual_outcomes: list) -> float:
        true_positives = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 1 and a == 1)
        actual_positives = sum(1 for a in actual_outcomes if a == 1)
        return true_positives / actual_positives if actual_positives > 0 else 0.0

    def calculate_f1_score(self, predictions: list, actual_outcomes: list) -> float:
        precision = self.calculate_precision(predictions, actual_outcomes)
        recall = self.calculate_recall(predictions, actual_outcomes)
        return 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0

    def generate_confusion_matrix(self, predictions: list, actual_outcomes: list) -> Dict:
        tp = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 1 and a == 1)
        tn = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 0 and a == 0)
        fp = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 1 and a == 0)
        fn = sum(1 for p, a in zip(predictions, actual_outcomes) if p == 0 and a == 1)

        return {
            'true_positives': tp,
            'true_negatives': tn,
            'false_positives': fp,
            'false_negatives': fn
        }

    def analyze_results(self, metrics: Dict) -> Dict:
        confusion_matrix = metrics['confusion_matrix']
        total = sum(confusion_matrix.values())

        return {
            'classification_distribution': {
                'positive_rate': (confusion_matrix['true_positives'] +
                                confusion_matrix['false_positives']) / total,
                'negative_rate': (confusion_matrix['true_negatives'] +
                                confusion_matrix['false_negatives']) / total
            },
            'error_analysis': {
                'false_positive_rate': confusion_matrix['false_positives'] /
                    (confusion_matrix['false_positives'] + confusion_matrix['true_negatives'])
                    if (confusion_matrix['false_positives'] + confusion_matrix['true_negatives']) > 0 else 0,
                'false_negative_rate': confusion_matrix['false_negatives'] /
                    (confusion_matrix['false_negatives'] + confusion_matrix['true_positives'])
                    if (confusion_matrix['false_negatives'] + confusion_matrix['true_positives']) > 0 else 0
            },
            'performance_by_category': {
                'accuracy': metrics['accuracy'],
                'precision': metrics['precision'],
                'recall': metrics['recall'],
                'f1_score': metrics['f1_score']
            }
        }

    def generate_recommendations(self, metrics: Dict) -> List[str]:
        recommendations = []

        # Accuracy-based recommendations
        if metrics['accuracy'] < 0.7:
            recommendations.append("Model accuracy is low - consider retraining with more data")
        elif metrics['accuracy'] < 0.85:
            recommendations.append("Model accuracy could be improved - consider fine-tuning")

        # Precision and Recall based recommendations
        if metrics['precision'] < 0.7:
            recommendations.append("High false positive rate - consider adjusting classification threshold")
        if metrics['recall'] < 0.7:
            recommendations.append("High false negative rate - consider lowering classification threshold")

        # Balance-based recommendations
        confusion_matrix = metrics['confusion_matrix']
        total_positives = confusion_matrix['true_positives'] + confusion_matrix['false_negatives']
        total_negatives = confusion_matrix['true_negatives'] + confusion_matrix['false_positives']
        if abs(total_positives - total_negatives) / (total_positives + total_negatives) > 0.2:
            recommendations.append("Dataset is imbalanced - consider using balanced training data")

        return recommendations

def test_classifier():
    predictions = [1, 0, 1, 1, 0, 1, 0, 0]
    actual_outcomes = [1, 0, 1, 0, 0, 1, 1, 0]

    evaluator = ClassificationEvaluator()

    results = evaluator.evaluate_classifications(predictions, actual_outcomes)

    print("\nClassification Metrics:")
    print(f"Accuracy: {results.metrics['accuracy']:.2f}")
    print(f"Precision: {results.metrics['precision']:.2f}")
    print(f"Recall: {results.metrics['recall']:.2f}")
    print(f"F1 Score: {results.metrics['f1_score']:.2f}")

    print("\nConfusion Matrix:")
    for key, value in results.metrics['confusion_matrix'].items():
        print(f"{key}: {value}")

    print("\nDetailed Analysis:")
    for category, metrics in results.detailed_analysis.items():
        print(f"\n{category}:")
        for metric, value in metrics.items():
            if isinstance(value, float):
                print(f"  {metric}: {value:.2f}")
            else:
                print(f"  {metric}: {value}")

    print("\nRecommendations:")
    for rec in results.recommendations:
        print(f"- {rec}")

    return results

if __name__ == "__main__":
    results = test_classifier()
```
